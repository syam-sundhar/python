```python
#IQR - Q3-Q1
#Q1-25
#Q3 - 75
#LB = Q1-1.5*IQR
#UB = Q3+1.5*IQR
```

```python
Q1 = df['Population in thousands (2017)'].quantile(0.25)
Q1
```

```python
Q3 = df['Population in thousands (2017)'].quantile(0.75)
Q3
```

```python
IQR = Q3-Q1
```

```python
LB = Q1-1.5*IQR
UB = Q3+1.5*IQR
```

```python
df = df[(df['Population in thousands (2017)'] >= LB) & (df['Population in thousands (2017)'] <= UB)]
```
